### Open AI Embedding

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
embeddings
```

```python
# single text embedding
single_text = "LangChain and RAG are awesome! Framework for building LLM applications."
single_embedding = embeddings.embed_query(single_text)
print(f"length of embedding: {len(single_embedding)}")
print(single_embedding)
```
