### Streaming And Batch

### Streaming
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.
Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:



Streaming means the model **returns output token by token** as it generates the response instead of waiting for the entire answer.

## Normal Response

```text
User:
Tell me about AI.

Wait...
Wait...
Wait...

AI stands for Artificial Intelligence...
```

## Streaming Response

```text
AI
stands
for
Artificial
Intelligence...
```

The user starts seeing the answer immediately instead of waiting for the complete response.

````python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] =os.getenv("GROQ_API_KEY")
````

````python
from langchain.chat_models import init_chat_model

llm = init_chat_model("qwen3:8b", model_provider="ollama")

llm
model = llm
````

````python
model.invoke("Write me a 200 words paragragh on Artificial Intelligence")  
# -------------------- Normal Response --------------------
#
# User:
# Tell me about AI.
#
# Wait...
# Wait...
# Wait...
#
# AI stands for Artificial Intelligence...
````

````python
prompt = """
Write a 200-word article on Artificial Intelligence.
"""

for chunk in model.stream(prompt):
    print(chunk.text, end="", flush=True)
````

````python
prompt = """
Why do parrots have colorful feathers?
"""

for chunk in model.stream(prompt):
    print(chunk.text, end="", flush=True)
````

````python
from IPython.display import display, Markdown, clear_output

response = ""

for chunk in model.stream("Write a 200-word article on Artificial Intelligence."):
    response += chunk.content
    clear_output(wait=True)
    display(Markdown(response))
````

#### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

# 2. Batch

Batch means **sending multiple inputs together** so the model processes them efficiently.

## Instead of

```text
Question 1
    │
    ▼
   LLM

Question 2
    │
    ▼
   LLM

Question 3
    │
    ▼
   LLM
```

You send all the inputs at once.

## Batch Processing

```text
Questions
[
    Q1,
    Q2,
    Q3
]
      │
      ▼
     LLM
      │
      ▼
Answers
[
    A1,
    A2,
    A3
]
```

````python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response)
````

````python
model.batch(
    ["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
````
