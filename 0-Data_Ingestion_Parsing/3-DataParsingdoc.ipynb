###  Word Document Processing

```python
from langchain_community.document_loaders import (
    Docx2txtLoader,
    UnstructuredWordDocumentLoader
)
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Methofd 1: Using Docx2txtLoader

print("Using Docx2txtLoader:")

try:
    docx_loader = Docx2txtLoader("data/docs/Rag.docx")
    docs = docx_loader.load()
    print(f"Loaded {len(docs)} document(s)")
    print(f"Content Preview: {docs[0].page_content[:200]}...")
    print(f"Metadata: {docs[0].metadata}")
    
    
    # chunking 
    print("\nChunking:")
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size = 500,
        chunk_overlap = 20
    )
    
    chunks = text_splitter.split_documents(docs)
    
    print(f"Created {len(chunks)}chunks")
    
    # preview first chunks
    print(f"\nChunk Preview:\n{chunks[0].page_content[:200]}...")
    print(f"\nMetadata:\n{chunks[0].metadata}")
    
except Exception as e:
    print(f"error loading docx file: {e}")
```

```python
## Method 2: Using UnstructuredWordDocumentLoader
print("\nUsing UnstructuredWordDocumentLoader:")


try:
    Unstructured_loader = UnstructuredWordDocumentLoader("data/docs/Rag.docx", mode="elements")
    docs = Unstructured_loader.load()
    print(f"Loaded {len(docs)} elements")
    for i, doc in enumerate(docs[:5]):  # Print first 3 elements
        print(f"\nElement {i+1}:")
        print(f"Type: {doc.metadata.get('category', 'Unknown')}")
        print(f"Content: {doc.page_content[:200]}...")
        
           # chunking 
    print("\nChunking:")
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size = 500,
        chunk_overlap = 20
    )
    
    chunks = text_splitter.split_documents(docs)
    
    print(f"Created {len(chunks)}chunks")
    
    # preview first chunks
    print(f"\nChunk Preview:\n{chunks[0].page_content[:200]}...")
    print(f"\nMetadata:\n{chunks[0].metadata}")
    
except Exception as e:
    print(f"error loading docx file: {e}")
```
